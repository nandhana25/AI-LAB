```python


state = [3, 1, 2, 0]

def calculate_cost(state):
    cost = 0

    for i in range(4):
        for j in range(i + 1, 4):

            if state[i] == state[j] or abs(state[i] - state[j]) == abs(i - j):
                cost += 1

    return cost


def generate_neighbors(state):
    neighbors = []

    for i in range(4):
        for j in range(i + 1, 4):

            new_state = state.copy()

            new_state[i], new_state[j] = new_state[j], new_state[i]

            neighbors.append((new_state, (i, j)))

    return neighbors


def hill_climbing(state):

    iteration = 0

    while True:

        current_cost = calculate_cost(state)

        print("\nIteration:", iteration)
        print("Current State:", state)
        print("Current Cost:", current_cost)

        if current_cost == 0:
            print("\nGoal State Reached!")
            break

        neighbors = generate_neighbors(state)

        print("\nNeighbor States:")

        best_state = None
        best_cost = float('inf')
        best_pair = None

        for neighbor, pair in neighbors:

            cost = calculate_cost(neighbor)

            print("Swap", pair, "->", neighbor, "Cost:", cost)

            if cost < best_cost:
                best_cost = cost
                best_state = neighbor
                best_pair = pair

        if best_cost >= current_cost:
            print("\nNo better neighbor found!")
            print("Hill Climbing stopped.")
            break

        print("\nSelected Swap:", best_pair)
        print("Next State:", best_state)
        print("Next Cost:", best_cost)

        state = best_state
        iteration += 1

    print("\nFinal State:", state)
    print("Final Cost:", calculate_cost(state))


hill_climbing(state)
```
